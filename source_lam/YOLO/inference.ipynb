```python
from ultralytics import YOLO
import os, glob, numpy as np

WEIGHTS_PATH = r'C:\Users\ADMIN\OneDrive\Documents\ViettelAIRace_Lam\source_lam\yolov11\train\models\weights\best.pt'  # model fine-tune của bạn
IMAGE_FOLDER = r'C:\Users\ADMIN\OneDrive\Documents\ViettelAIRace_Lam\dataset\Train\rgb'
OUTPUT_FOLDER = r'C:\Users\ADMIN\OneDrive\Documents\ViettelAIRace_Lam\source_lam\yolov11\train\models\masks'

os.makedirs(OUTPUT_FOLDER, exist_ok=True)
image_files = sorted(glob.glob(os.path.join(IMAGE_FOLDER, '*.png')))
model = YOLO(WEIGHTS_PATH)

total_saved = 0
for idx, img_path in enumerate(image_files, 1):
    print(f"[{idx}/{len(image_files)}] Xử lý: {os.path.basename(img_path)}")
    try:
        # ✅ Chỉ truyền path file
        results = model.predict(source=img_path, save=False, verbose=False)
        result = results[0]

        if result.masks is not None and result.masks.data is not None:
            masks_np = result.masks.data.cpu().numpy()
            for i, mask in enumerate(masks_np):
                out_path = os.path.join(
                    OUTPUT_FOLDER,
                    f"{os.path.splitext(os.path.basename(img_path))[0]}_mask{i}.npy"
                )
                np.save(out_path, mask)
                total_saved += 1
        else:
            print("  ⚠ Không có mask")
    except Exception as e:
        print(f"  ❌ Lỗi: {e}")

print(f"Tổng mask lưu được: {total_saved}")
```
